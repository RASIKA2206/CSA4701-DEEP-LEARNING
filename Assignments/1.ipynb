```python
import numpy as np
import matplotlib.pyplot as plt

# Student data: study hours, attendance, assignment score
X = np.array([
    [6, 90, 85],
    [4, 80, 75],
    [8, 95, 92]
])

# Weights
W = np.array([0.3, 0.3, 0.4])

# Matrix multiplication
output = X @ W

print("Student Data Matrix:")
print(X)
print("\nWeighted Outputs:")
print(output)

plt.bar(["Student 1", "Student 2", "Student 3"], output)
plt.title("Student Performance Weighted Output")
plt.ylabel("Weighted Score")
plt.show()
```
