```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy

# Actual label: 1 means spam
actual = 1

# Different predicted probabilities of spam
predictions = np.array([0.9, 0.6, 0.1])

# Binary cross-entropy
loss = -(
    actual * np.log(predictions)
    + (1 - actual) * np.log(1 - predictions)
)

for p, l in zip(predictions, loss):
    print(f"Prediction: {p:.1f}, Cross-Entropy Loss: {l:.3f}")

# Entropy of three risk categories
probabilities = [0.5, 0.3, 0.2]
print("\nClass Entropy:", round(entropy(probabilities), 3))

plt.bar(predictions.astype(str), loss)
plt.title("Cross-Entropy Loss")
plt.xlabel("Predicted Spam Probability")
plt.ylabel("Loss")
plt.show()
```
