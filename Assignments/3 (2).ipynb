```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Disease screening values
prevalence = 0.01
sensitivity = 0.95
false_positive = 0.05

# Bayes' theorem
positive_prob = (
    prevalence * sensitivity
) / (
    prevalence * sensitivity
    + (1 - prevalence) * false_positive
)

print("Probability of disease after positive test:")
print(round(positive_prob * 100, 2), "%")

# Normal distribution of waiting times
x = np.linspace(0, 60, 200)
y = norm.pdf(x, loc=30, scale=8)

plt.plot(x, y)
plt.title("Hospital Waiting Time Distribution")
plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Probability Density")
plt.show()
```
