```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Height": [160, 170, 165, 180, 175],
    "Weight": [55, 68, 60, 80, 72],
    "BP": [110, 120, 115, 135, 125]
})

cov = data.cov()

print("Covariance Matrix:")
print(cov.round(2))

values, vectors = np.linalg.eig(cov)

print("\nEigenvalues:")
print(np.round(values, 2))

print("\nEigenvectors:")
print(np.round(vectors, 2))

plt.imshow(data.corr(), cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(3), data.columns)
plt.yticks(range(3), data.columns)
plt.title("Feature Correlation Heatmap")
plt.show()
```
