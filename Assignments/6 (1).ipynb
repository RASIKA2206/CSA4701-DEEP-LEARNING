```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
X = np.array([1, 2, 3, 4], dtype=float)
y = np.array([3, 5, 7, 9], dtype=float)

# Initial parameters
m = 0.0
b = 0.0
learning_rate = 0.01
losses = []

# Gradient descent
for i in range(1000):
    predictions = m * X + b
    error = predictions - y

    dm = (2 / len(X)) * np.sum(error * X)
    db = (2 / len(X)) * np.sum(error)

    m -= learning_rate * dm
    b -= learning_rate * db

    losses.append(np.mean(error ** 2))

print("Final Slope:", round(m, 3))
print("Final Intercept:", round(b, 3))
print("Final Loss:", round(losses[-1], 6))

plt.plot(losses)
plt.title("Gradient Descent Loss Curve")
plt.xlabel("Iteration")
plt.ylabel("Mean Squared Error")
plt.show()
```
